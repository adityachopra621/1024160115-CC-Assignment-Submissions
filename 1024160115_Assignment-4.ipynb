```python
import pandas as pd
```

```python
ROLL_NO = 1024160115
NAME = "Aditya Chopra"

digits = [int(d) for d in str(ROLL_NO)[-2:]]

categories = ["billing", "account", "general"]

personalized_entries = []

# First personalized entry
for digit in digits:

    category = categories[digit % 3]

    if category == "account":
        entry = {
            "question": "How do I update my registered mobile number?",
            "answer": "Go to Account Settings and select Update Mobile Number.",
            "keywords": "mobile number update registered phone account",
            "category": category
        }

    elif category == "billing":
        entry = {
            "question": "How can I check my pending fee balance?",
            "answer": "Log in to the student portal and open the Fee Details section.",
            "keywords": "pending fee balance dues payment billing",
            "category": category
        }

    else:
        entry = {
            "question": "Where can I find the college academic calendar?",
            "answer": "Visit the college website and open the Academic Calendar section.",
            "keywords": "academic calendar college dates schedule general",
            "category": category
        }

    personalized_entries.append(entry)


print("Last two digits:", digits)
print("Personalized entries:")
for entry in personalized_entries:
    print(entry)
```

```python
fixed_entries = [

    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },

    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },

    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },

    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]
```

```python
all_entries = fixed_entries + personalized_entries

df = pd.DataFrame(all_entries)

print("Final Knowledge Base:")
display(df)

print("Total FAQ entries:", len(df))
```

```python
def score_query(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())

        matched_words = query_words.intersection(keywords)

        score = len(matched_words)

        if score > 0:

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "matched_keywords": list(matched_words),
                "score": score
            })

    results = sorted(
        results,
        key=lambda x: x["score"],
        reverse=True
    )

    return results


query = input("Enter your question: ")

matches = score_query(query, df)

print("\nMatching FAQ Entries:")

if matches:
    for result in matches:
        print("\nQuestion:", result["question"])
        print("Answer:", result["answer"])
        print("Category:", result["category"])
        print("Matched Keywords:", result["matched_keywords"])
        print("Confidence Score:", result["score"])
else:
    print("No matching FAQ found.")
```

```python
def same_category(category_name, df):

    result = df[
        df["category"].str.lower() == category_name.lower()
    ]

    return result[["question", "answer", "category"]]


personalized_category = personalized_entries[0]["category"]

print("Selected Category:", personalized_category)

print("\nFAQs belonging to this category:")

display(same_category(personalized_category, df))
```

```python
print("Available FAQ entries:")

for i, question in enumerate(df["question"]):
    print(i, question)


choice = int(input("\nEnter the index of the FAQ entry to update: "))

new_keyword = input("Enter a new keyword: ").strip().lower()

if 0 <= choice < len(df) and new_keyword:

    existing_keywords = df.loc[choice, "keywords"].split()

    if new_keyword not in existing_keywords:
        existing_keywords.append(new_keyword)

    df.loc[choice, "keywords"] = " ".join(existing_keywords)

    print("\nKeyword added successfully!")

else:
    print("Invalid choice or empty keyword.")


filename = f"{ROLL_NO}_faq_data.csv"

df.to_csv(filename, index=False)

print("Updated DataFrame:")
display(df)

print("File saved as:", filename)
```

```python
category_counts = df.groupby("category").size().reset_index(name="FAQ Count")

print("Number of FAQ entries in each category:")

display(category_counts)
```

```python
def score_query_with_ties(query, df):

    query_words = set(query.lower().split())

    results = []

    for index, row in df.iterrows():

        keywords = set(row["keywords"].lower().split())

        matched_words = query_words.intersection(keywords)

        score = len(matched_words)

        if score > 0:

            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "matched_keywords": list(matched_words),
                "score": score
            })

    if not results:
        print("No matching FAQ found.")
        return

    highest_score = max(result["score"] for result in results)

    top_matches = [
        result for result in results
        if result["score"] == highest_score
    ]

    print("Highest Score:", highest_score)

    if len(top_matches) > 1:
        print("Tie detected! All equally matching FAQs are displayed.")
    else:
        print("No tie. One FAQ has the highest score.")

    for result in top_matches:

        print("\nQuestion:", result["question"])
        print("Answer:", result["answer"])
        print("Category:", result["category"])
        print("Matched Keywords:", result["matched_keywords"])
        print("Score:", result["score"])
```

```python
print("DEMONSTRATION 1: QUERY WITH A TIE")

score_query_with_ties("fee", df)
```

```python
print("DEMONSTRATION 2: QUERY WITHOUT A TIE")

score_query_with_ties("password reset", df)
```
