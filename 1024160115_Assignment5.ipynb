```python
import numpy as np

# Create a 1-D array with 5 elements
arr = np.array([10, 20, 30, 40, 50])

print("Original array:", arr)

# a) Addition of 2
print("After adding 2:", arr + 2)

# b) Multiplication by 3
print("After multiplying by 3:", arr * 3)

# c) Division by 2
print("After dividing by 2:", arr / 2)
```

```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original array:", arr)
print("Reversed array:", arr[::-1])
```

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]

indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
import numpy as np

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]

indices = np.where(y == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

# a) 1st row, 2nd column
print("1st row, 2nd column:", arr[0, 1])

# b) 3rd row, 1st column
print("3rd row, 1st column:", arr[2, 0])
```

```python
import numpy as np

Aditya_Chopra = np.linspace(10, 100, 25)

print("Array:")
print(Aditya_Chopra)

print("Dimensions:", Aditya_Chopra.ndim)
print("Shape:", Aditya_Chopra.shape)
print("Total elements:", Aditya_Chopra.size)
print("Data type:", Aditya_Chopra.dtype)
print("Total bytes:", Aditya_Chopra.nbytes)

# Transpose using reshape()
transpose_array = Aditya_Chopra.reshape(1, 25)

print("Transpose using reshape():")
print(transpose_array)

# Transpose using T attribute
print("Transpose using T attribute:")
print(Aditya_Chopra.T)
```

```python
import numpy as np

ucs420_Aditya_Chopra = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Aditya_Chopra)

# Statistical operations
print("Mean:", np.mean(ucs420_Aditya_Chopra))
print("Median:", np.median(ucs420_Aditya_Chopra))
print("Maximum:", np.max(ucs420_Aditya_Chopra))
print("Minimum:", np.min(ucs420_Aditya_Chopra))
print("Unique elements:", np.unique(ucs420_Aditya_Chopra))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_Aditya_Chopra = ucs420_Aditya_Chopra.reshape(4, 3)

print("Reshaped Array (4 x 3):")
print(reshaped_ucs420_Aditya_Chopra)

# Resize to 2 rows and 3 columns
resized_ucs420_Aditya_Chopra = np.resize(ucs420_Aditya_Chopra, (2, 3))

print("Resized Array (2 x 3):")
print(resized_ucs420_Aditya_Chopra)
```
