```python
import re
import nltk
import matplotlib.pyplot as plt

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger_eng')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist
```

```python
paragraph = """
Artificial intelligence is transforming the modern world.
It helps people solve complex problems and automate daily tasks.
Machine learning allows computers to learn from data and improve their performance.
Artificial intelligence is widely used in healthcare, education, banking and transportation.
Technology makes communication faster and information more accessible.
However, responsible use of artificial intelligence is important for society.
"""

print(paragraph)
```

```python
clean_text = paragraph.lower()

clean_text = re.sub(r'[^a-zA-Z\s]', '', clean_text)

print("Lowercase text without punctuation:")
print(clean_text)
```

```python
sentences = sent_tokenize(paragraph)

words = word_tokenize(clean_text)

print("Sentence Tokenization:")
print(sentences)

print("\nWord Tokenization:")
print(words)
```

```python
split_words = clean_text.split()

nltk_words = word_tokenize(clean_text)

print("Using split():")
print(split_words)

print("\nUsing word_tokenize():")
print(nltk_words)

print("\nNumber of words using split():", len(split_words))
print("Number of words using NLTK:", len(nltk_words))
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in nltk_words
    if word not in stop_words
]

print("Words after removing stopwords:")
print(filtered_words)
```

```python
frequency = FreqDist(filtered_words)

print("Word Frequency Distribution:")
print(frequency.most_common())

frequency.plot(10, title="Top 10 Most Frequent Words")
plt.show()
```

```python
texts = [
    "Artificial intelligence is changing the world through smart technology.",
    "Machine learning helps computers learn from data and make predictions.",
    "Artificial intelligence and machine learning are useful in healthcare."
]

for i, text in enumerate(texts):
    print(f"Text {i+1}: {text}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer

count_vectorizer = CountVectorizer()

bow_matrix = count_vectorizer.fit_transform(texts)

print("Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

print("TF-IDF Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
feature_names = tfidf_vectorizer.get_feature_names_out()

for i in range(len(texts)):

    scores = tfidf_matrix[i].toarray()[0]

    top_indices = scores.argsort()[-3:][::-1]

    print(f"\nTop 3 keywords in Text {i+1}:")

    for index in top_indices:
        print(feature_names[index], ":", round(scores[index], 3))
```

```python
text1 = """
Artificial intelligence enables machines to perform tasks that normally require human intelligence.
It uses machine learning algorithms to analyze data and identify patterns.
AI is used in healthcare, education and automation.
It improves efficiency and supports decision making.
"""

text2 = """
Blockchain is a distributed technology that stores information securely.
It uses cryptographic methods to maintain transparent digital records.
Blockchain is widely used in finance and digital transactions.
It improves security, trust and data management.
"""
```

```python
from nltk.tokenize import word_tokenize

def preprocess(text):

    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)

    tokens = word_tokenize(text)

    stop_words = set(stopwords.words('english'))

    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    return tokens


tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("AI Tokens:")
print(tokens1)

print("\nBlockchain Tokens:")
print(tokens2)
```

```python
set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard_similarity = len(intersection) / len(union)

print("Common Words:", intersection)
print("Jaccard Similarity:", round(jaccard_similarity, 4))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform([text1, text2])

cosine_score = cosine_similarity(tfidf_matrix[0], tfidf_matrix[1])[0][0]

print("Cosine Similarity:", round(cosine_score, 4))
```

```python
print("Jaccard Similarity:", round(jaccard_similarity, 4))
print("Cosine Similarity:", round(cosine_score, 4))

if cosine_score > jaccard_similarity:
    print("\nCosine similarity gives higher similarity based on word importance.")
elif jaccard_similarity > cosine_score:
    print("\nJaccard similarity gives higher similarity based on common unique words.")
else:
    print("\nBoth metrics give equal similarity.")
```

```python
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt
```

```python
reviews = [
    "This laptop is excellent and has amazing performance.",
    "I absolutely love this product. The quality is fantastic.",
    "The battery life is great and the design is beautiful.",
    "This product is terrible and very disappointing.",
    "The laptop is slow and the battery is poor.",
    "I am unhappy with the quality and performance.",
    "The product is okay and works as expected.",
    "The service is average and nothing special."
]
```

```python
for review in reviews:

    analysis = TextBlob(review)

    print("\nReview:", review)
    print("Polarity:", analysis.sentiment.polarity)
    print("Subjectivity:", analysis.sentiment.subjectivity)
```

```python
positive_reviews = []
negative_reviews = []
neutral_reviews = []

for review in reviews:

    polarity = TextBlob(review).sentiment.polarity

    if polarity > 0:
        positive_reviews.append(review)

    elif polarity < 0:
        negative_reviews.append(review)

    else:
        neutral_reviews.append(review)


print("Positive Reviews:", positive_reviews)
print("\nNegative Reviews:", negative_reviews)
print("\nNeutral Reviews:", neutral_reviews)
```

```python
positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color='white'
).generate(positive_text)

plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Word Cloud of Positive Reviews")
plt.show()
```

```python
training_text = """
Artificial intelligence is an important part of modern technology.
It allows computers to perform tasks that usually require human intelligence.
Machine learning helps computers learn from examples and improve their performance.
Deep learning uses neural networks to understand complex patterns in data.
Artificial intelligence is used in healthcare, education, banking and transportation.
In healthcare, intelligent systems help doctors analyze medical information.
In education, technology supports personalized learning for students.
In banking, machine learning helps detect fraud and improve security.
Natural language processing allows computers to understand and generate human language.
These technologies are changing industries and creating new opportunities.
However, artificial intelligence must be developed responsibly.
Data privacy, fairness and transparency are important considerations.
In the future, intelligent systems may become more useful in everyday life.
Human creativity and decision making will continue to play an important role.
Technology should support people and improve the quality of life.
"""
print("Training Data:")
print(training_text)

print("\nWord Count:", len(training_text.split()))
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

tokenizer = Tokenizer()

tokenizer.fit_on_texts([training_text])

total_words = len(tokenizer.word_index) + 1

print("Vocabulary Size:", total_words)

input_sequences = []

for line in training_text.split('.'):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)

print("\nTotal Input Sequences:", len(input_sequences))
print("\nSample Sequences:")
print(input_sequences[:5])
```

```python
import numpy as np

max_sequence_len = max(len(seq) for seq in input_sequences)

padded_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding='pre'
)

X = padded_sequences[:, :-1]
y = padded_sequences[:, -1]

print("Input Shape:", X.shape)
print("Output Shape:", y.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

model = Sequential([
    Embedding(total_words, 32),
    LSTM(64),
    Dense(total_words, activation='softmax')
])

model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

model.summary()
```

```python
history = model.fit(
    X,
    y,
    epochs=150,
    verbose=1
)
```

```python
def generate_text(seed_text, next_words):

    for _ in range(next_words):

        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding='pre'
        )

        predicted = model.predict(token_list, verbose=0)

        predicted_word_index = np.argmax(predicted, axis=1)[0]

        output_word = ""

        for word, index in tokenizer.word_index.items():
            if index == predicted_word_index:
                output_word = word
                break

        if not output_word:
            break

        seed_text += " " + output_word

    return seed_text


seed_text = "artificial intelligence"

generated_text = generate_text(seed_text, 25)

print("Generated Text:")
print(generated_text)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(history.history['accuracy'])

plt.title("LSTM Model Training Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")

plt.show()
```
